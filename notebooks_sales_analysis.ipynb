```python

import sqlite3

conn = sqlite3.connect("sales_analysis.db")

conn.execute("""
CREATE TABLE IF NOT EXISTS sales (
    order_id INTEGER PRIMARY KEY,
    product TEXT,
    category TEXT,
    quantity INTEGER,
    price REAL
)
""")

print("Sales database created successfully!")
```

```python
sales_data = [
    (1, "Laptop", "Electronics", 2, 45000),
    (2, "Mouse", "Accessories", 10, 500),
    (3, "Keyboard", "Accessories", 5, 1200),
    (4, "Monitor", "Electronics", 3, 12000),
    (5, "Headphones", "Accessories", 4, 2000),
    (6, "Laptop", "Electronics", 1, 45000),
    (7, "Monitor", "Electronics", 2, 12000),
    (8, "Mouse", "Accessories", 8, 500),
    (9, "Keyboard", "Accessories", 6, 1200),
    (10, "Headphones", "Accessories", 3, 2000)
]

conn.executemany("""
INSERT OR REPLACE INTO sales
(order_id, product, category, quantity, price)
VALUES (?, ?, ?, ?, ?)
""", sales_data)

conn.commit()

print("10 sales records added successfully!")
```

```python
import sqlite3

conn = sqlite3.connect(":memory:")

conn.execute("""
CREATE TABLE sales (
    order_id INTEGER PRIMARY KEY,
    product TEXT,
    category TEXT,
    quantity INTEGER,
    price REAL
)
""")

print("New database created successfully!")
```

```python
sales_data = [
    (1, "Laptop", "Electronics", 2, 45000),
    (2, "Mouse", "Accessories", 10, 500),
    (3, "Keyboard", "Accessories", 5, 1200),
    (4, "Monitor", "Electronics", 3, 12000),
    (5, "Headphones", "Accessories", 4, 2000),
    (6, "Laptop", "Electronics", 1, 45000),
    (7, "Monitor", "Electronics", 2, 12000),
    (8, "Mouse", "Accessories", 8, 500),
    (9, "Keyboard", "Accessories", 6, 1200),
    (10, "Headphones", "Accessories", 3, 2000)
]

conn.executemany("""
INSERT INTO sales
VALUES (?, ?, ?, ?, ?)
""", sales_data)

conn.commit()

print("10 sales records added successfully!")
```

```python
result = conn.execute("""
SELECT
    SUM(quantity) AS total_quantity,
    SUM(quantity * price) AS total_revenue
FROM sales
""")

row = result.fetchone()

print("Total Quantity Sold:", row[0])
print("Total Revenue: Rs.", row[1])
```

```python
result = conn.execute("""
SELECT
    product,
    SUM(quantity) AS units_sold,
    SUM(quantity * price) AS revenue
FROM sales
GROUP BY product
ORDER BY units_sold DESC
""")

for row in result.fetchall():
    print(row)
```

```python

result = conn.execute("""
SELECT
    COUNT(*) AS total_records,
    COUNT(DISTINCT order_id) AS unique_orders
FROM sales
""")

print(result.fetchone())
```

```python
result = conn.execute("""
SELECT
    category,
    SUM(quantity) AS units_sold,
    SUM(quantity * price) AS revenue
FROM sales
GROUP BY category
ORDER BY revenue DESC
""")

for row in result.fetchall():
    print(row)
```

```python
result = conn.execute("""
SELECT
    COUNT(*) AS total_orders,
    COUNT(DISTINCT product) AS products,
    SUM(quantity * price) AS total_revenue
FROM sales
""")

print("PROJECT SUMMARY")
print("----------------")
print("Total Orders:", result.fetchone()[0])

result = conn.execute("SELECT COUNT(DISTINCT product) FROM sales")
print("Unique Products:", result.fetchone()[0])

result = conn.execute("SELECT SUM(quantity * price) FROM sales")
print("Total Revenue: Rs.", result.fetchone()[0])
```
